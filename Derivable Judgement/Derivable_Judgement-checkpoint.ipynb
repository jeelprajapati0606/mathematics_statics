## 2. Confidence Intervals for key numerical data like age

```python
import numpy as np

ages = df['age'].values

# Mean and Standard Deviation
mean_a = np.mean(ages)
std_a = np.std(ages, ddof=1)
n = len(ages)

# Standard Error
se_a = std_a / np.sqrt(n)

# 95% Confidence Interval (z = 1.96)
z = 1.96
lower_a = mean_a - z * se_a
upper_a = mean_a + z * se_a

print("Mean Age:", mean_a)
print("95% CI for Age:", (lower_a, upper_a))



```

### Interpretation:
The average Age is around 47 years.  
The 95% confidence interval is [44, 49].  
This means the true average wAge of the population lies in this range.

# 3. Critical Value and p-value

```python

# Smokers vs Non-Smokers BMI


smokers = df[df['smoking_status']=="Smoker"]['bmi'].values
nonsmokers = df[df['smoking_status']=="Non-Smoker"]['bmi'].values

# Sample means
S1 = np.mean(smokers)
S2 = np.mean(nonsmokers)

# Sample standard deviations
sigma1 = np.std(smokers, ddof=1)
sigma2 = np.std(nonsmokers, ddof=1)

# Sample sizes
n1 = len(smokers)
n2 = len(nonsmokers)

alpha = 0.05

# z-score
z_score = (S1 - S2) / np.sqrt((sigma1**2/n1) + (sigma2**2/n2))

# p-value (two-tailed)
p_value = 2 * (1 - norm.cdf(abs(z_score)))

# critical value for 95% confidence
critical_z = norm.ppf(1 - alpha/2)

print("Critical Value:", critical_z)
print("p-value:", p_value)
```

### Interpretation (Smokers vs Non-Smokers BMI)
Critical Value: 1.9599

p-value: 0.6171

Since the p-value (0.6171) > 0.05 and the test statistic did not exceed the critical value, we fail to reject H₀.

👉 This means there is no statistically significant difference in BMI between Smokers and Non-Smokers in your dataset.

## 5. chi-square test on categorical data (smoking status and diabetes)

```python
# Step 1: Observed Frequencies

# Contingency table: Smoking vs Diabetes

table = pd.crosstab(df['smoking_status'], df['diabetes'])


print(table)

# Observed frequencies
observed = table.values
```

```python
# Step 2: Expected Frequencies

row_totals = observed.sum(axis=1).reshape(-1,1)
col_totals = observed.sum(axis=0).reshape(1,-1)
grand_total = observed.sum()

expected = (row_totals @ col_totals) / grand_total
print("Expected Frequencies:\n", expected)
```

```python
# Step 3: Chi-Square Statistic

chi_square_stat = ((observed - expected)**2 / expected).sum()

```

```python
# Step 4: Degrees of Freedom

df_chi = (observed.shape[0] - 1) * (observed.shape[1] - 1)
```

```python
# Step 5: Critical Value & p-value

alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, df_chi)
p_value = 1 - stats.chi2.cdf(chi_square_stat, df_chi)
```

```python
#  Step 6: Decision (from p_value or from critical_value)

if p_value < alpha:
    print("Reject the null hypothesis: There is a significant association between smoking status and diabetes.")

else:
    print("Fail to reject the null hypothesis: There is no significant association between smoking status and diabetes.")
```

```python
#  Final Output

print("Chi-Square Statistic:", chi_square_stat),
print("Degrees of Freedom:", df_chi),
print("Critical Value:", critical_value),
print("p-value:", p_value),

```

## 6. ANOVA test to check if age groups significantly differ in disease rate

```python
# Step 1: Prepare Groups



# Split data by age groups (Hypertension as disease rate)
group_18_25 = df[df['age_group']=="18-25"]['hypertension'].values
group_26_35 = df[df['age_group']=="26-35"]['hypertension'].values
group_36_45 = df[df['age_group']=="36-45"]['hypertension'].values
group_46_60 = df[df['age_group']=="46-60"]['hypertension'].values
group_60_plus = df[df['age_group']=="60+"]['hypertension'].values

groups = [group_18_25, group_26_35, group_36_45, group_46_60, group_60_plus]
```

```python
# Step 2: Calculate Total observations and Grand Mean

n = (len(group_18_25) + len(group_26_35) + len(group_36_45) +
     len(group_46_60) + len(group_60_plus))
k = 5

grand_mean = (sum(group_18_25) + sum(group_26_35) + sum(group_36_45) +
              sum(group_46_60) + sum(group_60_plus)) / n
```

```python
# step 3: group means 

mean_A = np.mean(group_18_25)
mean_B = np.mean(group_26_35)
mean_C = np.mean(group_36_45)
mean_D = np.mean(group_46_60)
mean_E = np.mean(group_60_plus)
```

```python
# Step 4: Between Group Sum of Squares (SSB)

SSB = (
    len(group_18_25) * (mean_A - grand_mean)**2 +
    len(group_26_35) * (mean_B - grand_mean)**2 +
    len(group_36_45) * (mean_C - grand_mean)**2 +
    len(group_46_60) * (mean_D - grand_mean)**2 +
    len(group_60_plus) * (mean_E - grand_mean)**2
)
```

```python
# Step 5: Within Group Sum of Squares (SSW)

SSW = 0
for x in group_18_25: SSW += (x - mean_A)**2
for x in group_26_35: SSW += (x - mean_B)**2
for x in group_36_45: SSW += (x - mean_C)**2
for x in group_46_60: SSW += (x - mean_D)**2
for x in group_60_plus: SSW += (x - mean_E)**2
```

```python
# Step 6: Degrees of Freedom

df_between = k - 1
df_within = n - k
```

```python
# Step 7: Mean Squares

MSB = SSB / df_between
MSW = SSW / df_within
```

```python
# Step 8: F‑Statistic

F_static = MSB / MSW 
print("F-Statistic:", F_static)
```

```python
#  Critical F-value



alpha = 0.05
critical_F = stats.f.ppf(1 - alpha, df_between, df_within)

print("Critical Value (F):", critical_F)
```

```python
# p‑value approach

alpha = 0.05

if p_value < alpha:
    print("The p-value is less than 0.05, so we reject the null hypothesis. At least one age group differs significantly in disease rate.")
else:
    print("The p-value is greater than or equal to 0.05, so we fail to reject the null hypothesis. No significant difference among age groups.")

```

```python
# critical value approach

if F_static > critical_F:
    print("The F-statistic falls outside the decision boundary, so we reject the null hypothesis. At least one age group differs significantly in disease rate.")
else:
    print("The F-statistic falls within the decision boundary, so we fail to reject the null hypothesis. Age groups perform similarly in disease rate.")
```

### Interpretation (Age Groups vs Disease Rate)
F-Statistic: 2.2644

Critical Value (F): 2.4179

p-value ≥ 0.05

Since the p-value ≥ 0.05 and the F-statistic (2.2644) < Critical Value (2.4179), we fail to reject H₀.

👉 This means there is no statistically significant difference in disease rate among the age groups in your dataset.

👉 In other words, age groups perform similarly in disease rate.

## 7. Covariance and Correlation between continuous variables (e.g., Age vs. BMI)

```python
# Step 1: Extract Variables

age = df['age'].values
bmi = df['bmi'].values

# Step 2: Calculate Means

mean_age = np.mean(age)
mean_bmi = np.mean(bmi)

# Step 3: Covariance

covariance = np.sum((age - mean_age) * (bmi - mean_bmi)) / (len(age) - 1)
print("Covariance(Age VS BMI):", covariance)

# Step 4: Correlation

std_age = np.std(age, ddof=1)
std_bmi = np.std(bmi, ddof=1)

Correlation = covariance / (std_age * std_bmi)
print("Correlation (Age vs BMI):", Correlation)


```
